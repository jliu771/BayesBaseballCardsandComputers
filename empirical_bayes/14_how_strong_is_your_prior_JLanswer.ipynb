## 1. Spread is strength

With mean μ, strength ν = α + β and standard deviation σ:

$$\nu = \frac{\mu(1-\mu)}{\sigma^2} - 1, \qquad \alpha = \mu\nu, \qquad \beta = (1-\mu)\nu.$$

**14.1.1 · Your turn.** Write `strength_from_spread(mu, sigma)`. With μ = 0.27, make a table of ν, α and β for σ = 0.02, 0.0256, 0.035 and 0.05.

````python
# 14.1.1 · Your code here
def strength_from_spread(mu, sigma):
    alpha = ((mu ** 2) * (1 - mu) / (sigma ** 2)) - mu
    beta = ((mu * ((1 - mu) ** 2))/ (sigma ** 2)) - (1 - mu)
    return alpha, beta

m = 0.27
sigmas = [0.02, 0.0256, 0.035, 0.05]
hypers = []

for sig in sigmas:
    a, b, = strength_from_spread(m, sig)
    hypers.append({'Mu': m, 'Sigma': sig, 'Alpha': a, 'Beta': b, 'Spread': a+b})
    
hypers = pd.DataFrame(hypers)
hypers
````

> **Check:** σ = 0.0256 gives about Beta(81, 219), ν ≈ 300; σ = 0.02 gives ν ≈ 490; σ = 0.035 gives ν ≈ 160.

**14.1.2 · Your turn.** Plot the three priors for σ = 0.02, 0.0256 and 0.035 on one chart.

````python
# 14.1.2 · Your code here

x = np.linspace(0.15, 0.4, 200)
for i, row in hypers.iterrows():
    plt.plot(x, stats.beta(row.Alpha, row.Beta).pdf(x), label=f"σ = {row.Sigma:.4f}  (α+β = {row.Alpha + row.Beta:.0f})")
plt.xlabel("Batting averages")
plt.ylabel("Density")
plt.legend(bbox_to_anchor = (1.05, 0.5), loc = 'center left', title = 'Prior')
plt.show()
````

## 2. What strength does to an estimate

The posterior mean is a weighted average of the player's record and the prior mean:

$$\hat p = \frac{H + \mu\nu}{AB + \nu} = w\,\frac{H}{AB} + (1-w)\,\mu, \qquad w = \frac{AB}{AB+\nu}.$$

**14.2.1 · Your turn.** Plot the weight w against AB (log x-axis, AB from 1 to 20,000) for ν = 160, 300 and 490.

````python
# 14.2.1 · Your code here
nus = [160, 300, 490]
ab = np.logspace(0, np.log10(20000), num = 200, base = 10.0)

for nu in nus:
    w = ab / (ab + nu)
    plt.plot(ab, w, label = f"Nu: {nu}")

plt.xlabel("AB, log scale")
plt.xscale("log")
plt.ylabel("Weight w")
plt.show()
````

**14.2.2 · Your turn.** For records 4/10, 100/300 and 3000/10000, tabulate the posterior mean under each of the three priors.

````python
# 14.2.2 · Your code here
posts = pd.concat([hypers] * 3, ignore_index = True)
posts['H'] = np.repeat([4, 100, 3000], 4)
posts['AB'] = np.repeat([10, 300, 10000], 4)
posts['mean_post'] = (posts.Alpha + posts.H) / (posts.Alpha + posts.Beta + posts.AB)
print(posts)
````

````python
import seaborn as sns
sc = sns.scatterplot(data = posts, x = 'AB', y = 'mean_post', hue = 'Spread', palette = 'viridis')
plt.xlabel("# Simulated ABs")
plt.xscale("log")
plt.ylabel("Estimated Batting Average")
sns.move_legend(sc, "upper left", bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
````

## 3. Let the data decide: baseball

Use `eb_data.career(pitcher_rule=None)` and, as in Lesson 02, fit on players with AB > 500. Write the beta-binomial with mean μ and strength ν: `stats.betabinom.logpmf(H, AB, mu * nu, (1 - mu) * nu)`.

**14.3.1 · Your turn.** Write `fit_mu_nu(H, AB)` that finds the maximum-likelihood μ and ν. Optimizing `logit(mu)` and `log(nu)` keeps them in range. Report μ, ν, α, β and the implied σ.

````python
# 14.3.1 · Your code here
career = eb_data.career(pitcher_rule = None)
career500 = career[career.AB > 500]

def fit_mu_nu(H, AB):
    
    def neg_log_lik_mu_nu(params, H, AB):
        logit_m, log_n= params
        m, n = special.expit(logit_m), np.exp(log_n)
        return -stats.betabinom.logpmf(H, AB, (m * n), (1-m)*n).sum()
        
    res = optimize.minimize(neg_log_lik_mu_nu, x0 =[special.logit(0.26), np.log(300)] , args = (H, AB), method = "L-BFGS-B")
    mu, nu = res.x
    return special.expit(mu), np.exp(nu)

mu0, nu0 = fit_mu_nu(career500.H, career500.AB)
print(f"Fits: mu: {mu0:.3f}, nu: {nu0:.1f}, alpha: {mu0 * nu0:.2f}, beta: {(1-mu0) * nu0:.2f}, sigma: {np.sqrt(mu0 * (1 - mu0) / nu0):.3f}")
````

> **Check:** with the 2025 Lahman data, μ ≈ 0.260 and ν ≈ 367 (about Beta(95, 271)), implied σ ≈ 0.023.
